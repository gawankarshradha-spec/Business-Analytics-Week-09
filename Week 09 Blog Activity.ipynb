```python
# Read the employee CSV file into a pandas DataFrame.
employees_df = pd.read_csv('hr_employees.csv')

# Save the data as the employees table using the existing connection.
# Replace the table if it exists and exclude the pandas row index.
employees_df.to_sql('employees', conn, if_exists='replace', index=False)

# Find employees in Chicago with more than 10 years of experience.
# SELECT * retrieves all columns.
# AND requires both conditions to be true.
result = pd.read_sql_query('''
SELECT *
FROM employees
WHERE city = 'Chicago'
  AND years_experience > 10
''', conn)

# Display the matching employee records.
display(result)
```

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load restaurant_orders.csv into the 'orders' table
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# GROUP BY combines records into groups for summary calculations.
# SUM() adds values.
# AS gives a result column or table a short name.

# Student Practice: Total impressions for each week
pd.read_sql_query('''
SELECT week, SUM(impressions) AS total_impressions
FROM campaigns
GROUP BY week
''', conn)
```

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load restaurant_orders.csv into the 'orders' table
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# SUM() calculates the total sales for each date.
# GROUP BY organizes records by date.
# HAVING filters dates where total sales exceeded $400.
# AS gives the calculated column a name.

# Student Practice: Find dates with total sales above $400

pd.read_sql_query('''
SELECT order_date, SUM(order_total) AS total_sales
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load clinic_visits.csv into the 'visits' table
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# GROUP BY combines records into groups for summary calculations.
# AVG() calculates the average of nonmissing values.
# ROUND(value, digits) rounds a number to the specified decimal places.
# MAX() returns the largest nonmissing value.
# AS gives a result column or table a short name.

# Student Practice: Patient age and waiting time by doctor
pd.read_sql_query('''
SELECT doctor,
       ROUND(AVG(patient_age), 1) AS avg_patient_age,
       MAX(wait_minutes) AS longest_wait
FROM visits
GROUP BY doctor
''', conn)
```

  <p>
    Using the <strong>customers</strong> and
    <strong>transactions</strong> tables, write a SQL query
    to complete the following tasks:
  </p>

  <ol>
    <li>Join both tables using <code>customer_id</code>.</li>
    <li>Calculate the total transaction <code>amount</code> for each <code>branch</code> using <code>SUM()</code>.</li>
    <li>Use <code>GROUP BY</code> to organize the results by branch.</li>
    <li>Run the query using <code>pd.read_sql_query()</code> with <code>conn</code>.</li>
    <li>Display the results with one row per branch.</li>

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load both files into their own tables
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# AS gives a result column or table a short name.
# JOIN matches rows from tables using the ON condition.
# SUM() adds values.
# GROUP BY combines records into groups for summary calculations.

# Student Practice: Total transaction amount for each branch
pd.read_sql_query('''
SELECT c.branch, SUM(t.amount) AS total_amount
FROM transactions AS t
JOIN customers AS c
ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
